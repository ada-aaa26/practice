Now, let's change the weights so that the filter is an "edge detector", and then apply the filter.

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

# Horizontal edge detector: top row positive, bottom row negative
horizontal_filter = torch.tensor([[ 1.,  1.,  1.],
                                  [ 0.,  0.,  0.],
                                  [-1., -1., -1.]], dtype=torch.float32)

# Conv2d weights have shape (out_channels, in_channels, kH, kW) → (1, 1, 3, 3)
conv = nn.Conv2d(in_channels=1, out_channels=1, kernel_size=3, bias=False)
with torch.no_grad():
    conv.weight.copy_(horizontal_filter.view(1, 1, 3, 3))

# Conv2d expects input of shape (batch, channels, H, W)
img = torch.as_tensor(x, dtype=torch.float32)
if img.dim() == 2:                # (H, W) → (1, 1, H, W)
    img = img.unsqueeze(0).unsqueeze(0)

out = conv(img)

# Plot the original and the filtered image side by side
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(img.squeeze())
axes[0].set_title('Original')
axes[1].imshow(out.detach().squeeze())
axes[1].set_title('Horizontal edges')
plt.show()
```

```python
# TODO: change the settings for the convolutional layer above and predict how the output shape will change
# How does the output shape change with `kernel_size`, `stride`, `padding`, and `dilation`?

conv = nn.Conv2d(in_channels = 1, out_channels = 1, kernel_size = 3, bias = False)
y = conv(x.reshape(1,256,256))
print(y.shape)
# how does changing kernel_size = 5 change the final shape?

conv = nn.Conv2d(in_channels = 1, out_channels = 1, kernel_size = 5, bias = False)
y = conv(x.reshape(1,256,256))
print(y.shape)

# How does padding change the shape of the output channel? it doesn't change

conv = nn.Conv2d(in_channels = 1, out_channels = 1, kernel_size = 3, bias = False, padding = 'same')
y = conv(x.reshape(1,256,256))
print(y.shape)

# what happens if we use stride?
conv = nn.Conv2d(in_channels = 1, out_channels = 1, kernel_size = 5, bias = False, stride = 2)
y = conv(x.reshape(1,256,256))
print(y.shape)
```
